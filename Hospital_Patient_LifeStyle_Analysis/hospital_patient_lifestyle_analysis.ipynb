Healthcare organizations need data-driven insights to identify high-risk groups, understand lifestyle patterns, and support preventive healthcare strategies.

### Business Question 1

Does age influence heart attack risk?

Age is one of the most important demographic factors. Analyzing age can help identify whether certain age groups are more prone to heart attack risk.

```python
import seaborn as sns

sns.boxplot(data=df,x="Heart Attack Risk",y="Age",color="skyblue")
plt.figure(figsize=(8,5))
plt.tight_layout()
plt.show()
```

## Observation:

The age distribution is similar for both heart attack risk groups. The median age and overall spread do not show a noticeable difference, suggesting that age alone does not appear to be strongly associated with heart attack risk in this dataset.


### Business Question 2

Does smoking increase heart attack risk?

This analysis compares the number of patients with and without heart attack risk based on their smoking status.

```python


plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="Smoking",
    hue="Heart Attack Risk",
    palette="Set2"
)

plt.title("Smoking vs Heart Attack Risk")
plt.xlabel("Smoking Status")
plt.ylabel("Number of Patients")
plt.legend(title="Heart Attack Risk")
plt.tight_layout()
plt.savefig("Smoking_vs_Heart_attack_risk.png")
plt.show()
```

## Observation: 

Observation: The dataset contains considerably more smokers than non-smokers. In both smoking groups, the number of patients without heart attack risk is higher than the number with heart attack risk. Since the number of smokers and non-smokers is unequal, this graph alone is not sufficient to conclude whether smoking is associated with a higher heart attack risk.

### Business Question 3

Does cholesterol increase heart attack risk?

Cholesterol is one of the major health indicators associated with cardiovascular diseases. This analysis compares cholesterol levels between patients with and without heart attack risk.

```python
plt.figure(figsize=(6,5))

sns.boxplot(
    data=df,
    x="Heart Attack Risk",
    y="Cholesterol",
    color="skyblue"
    
)

plt.title("Cholesterol vs Heart Attack Risk")
plt.xlabel("Heart Attack Risk")
plt.ylabel("Cholesterol")
plt.tight_layout()
plt.show()
```

## Observation:

Patients with heart attack risk have a slightly higher median cholesterol level than patients without heart attack risk. However, the difference is small, suggesting that cholesterol alone may not strongly distinguish the two groups in this dataset.

## Correlation Analysis

Correlation analysis helps identify the strength and direction of relationships between numerical variables in the dataset. A heatmap is used to visualize the correlation matrix.

```python
# Select only numerical columns
numerical_df = df.select_dtypes(include=['number'])

# Compute correlation matrix
corr_matrix = numerical_df.corr()

# Plot heatmap
plt.figure(figsize=(12,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")
plt.savefig("Correlation_heatmap.png")
plt.show()
```

### Correlation Analysis - Observation

- Most numerical variables show very weak correlations (close to 0), indicating little or no linear relationship between them.
- Age and Smoking have a moderate positive correlation (0.39), which is the strongest relationship observed in the dataset.
- Variables such as BMI, Cholesterol, Stress Level, Exercise Hours, and Sleep Hours do not exhibit strong correlations with other numerical variables.
- Overall, the dataset shows limited linear associations among the numerical features.

# Key Findings

1. Most numerical variables showed weak correlations with each other, indicating limited linear relationships.

2. Cholesterol, BMI, and Exercise Hours showed only slight differences between patients with and without heart attack risk.

3. Lifestyle factors such as Smoking, Diet, and Diabetes did not show a clear visual separation between heart attack risk groups.
